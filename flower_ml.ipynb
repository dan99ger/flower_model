```python
import pandas as pd
url ="https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
df=pd.read_csv(url, names=columns)
#1.معرفة ابعاد مصفوفة البيانات
print("أبعاد مصفوفة البيانات:", df.shape)
#2.عرض انواع البيانات وهل توجد قيم مفقودة
print("\nأنواع البيانات:\n", df.info())
#3.الاحصاءات الوصفية للبيانات
print("\nالإحصاءات الوصفية للبيانات:\n", df.describe())
```

```python
#فحص وجود قيم مفقودة
missing_values = df.isnull().sum()
print("\nعدد القيم المفقودة في كل عمود:\n", missing_values)
#التعويض عن القيم المفقودة باستخدام المتوسط
df['sepal_length'].fillna(df['sepal_length'].mean(), inplace=True) 
```

```python
from sklearn.preprocessing import LabelEncoder
#تحويل القيم النصية إلى قيم رقمية
le = LabelEncoder()
df['species_encoded'] = le.fit_transform(df['species'])
print("\nالبيانات بعد تحويل القيم النصية إلى قيم رقمية:\n")
print(df[['species','species_encoded']].drop_duplicates())
```

```python
#تحتوي على الميزات الحسابية فقط x
x=df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
#تحتوي على المتغير الهدف y
y=df['species_encoded'] 
```

```python
from sklearn.preprocessing import StandardScaler
#توحيد البيانات 
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

print("\nشكل الميزات بعد التوحيد:\n", x_scaled[:5])  # عرض أول 5 صفوف من البيانات بعد التوحيد
```

```python
from sklearn.model_selection import train_test_split
# تقسيم البيانات إلى مجموعة التدريب والاختبار   
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.2, random_state=42)
print(f"\nحجم بيانات التدريب : {x_train.shape}, حجم بيانات الاختبار: {x_test.shape}")
```

```python
from sklearn.ensemble import RandomForestClassifier
#انشاء كائن النموذج
model= RandomForestClassifier(n_estimators=100,random_state=42)
#انشاءالنموذج على بيانات التدريب 
model.fit(x_train,y_train)
print("\nتم تدريب النموذج بنجاح!")
```

```python
#التنبؤ على بيانات الاختبار
y_pred = model.predict(x_test)

print("أول 5 تنبؤات للنموذج:\n", y_pred[:5])
print("أول 5 قيم حقيقية فعلية:\n", y_test.values[:5])
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
# حساب الدقة    
accuracy = accuracy_score(y_test, y_pred)
print(f"(Accuracy)دقة النموذج: {accuracy * 100:.2f}%\n")
 #عرض تقرير التصنيف
print("تقرير التصنيف:\n", classification_report(y_test, y_pred,target_names=le.classes_))
#عرض مصفوفةالارتباك
conf_matrix = confusion_matrix(y_test, y_pred)  
print("مصفوفة الارتباك:\n", conf_matrix)
```
